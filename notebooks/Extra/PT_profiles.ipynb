```python
import os
import numpy as np

import matplotlib as mpl
mpl.rcParams['figure.dpi'] = 300
from matplotlib.pyplot import figure, show, subplots
from matplotlib import pyplot as plt
from mpl_toolkits.axes_grid1.inset_locator import inset_axes
import netCDF4 as nc
```

```python
# AGNI pt profiles
AGNI_out = "/dataserver/users/formingworlds/borgmann/output/AGNI/raw/" #AGNI_spectrum_-4.0_0.1/atm.nc

fig = figure(figsize=(6,6))
frame = fig.add_subplot(1,1,1)
for folder in os.listdir(AGNI_out):
    if folder in ["AGNI_spectrum_4.0_1.0","AGNI_spectrum_4.0_5.0"]:
        continue
    path = AGNI_out + folder + "/atm.nc"

    ds = nc.Dataset(path)
    pressure = ds['p'][:]
    temp = ds['tmp'][:]
    frame.plot(temp,pressure[::-1]*10**-5,label=folder)
    
frame.set_yscale('log')
frame.set_xlabel("Temperature (K)")
frame.set_ylabel("Pressure (bar)")
frame.legend(loc="upper right")
show()
```

```python
# ARCiS pt profiles

ARCiS_path = "/dataserver/users/formingworlds/borgmann/ARCiS_forwardmodels/" #/fm_ARCiS-4.0_0.1_5/mixingratios.dat"

fig = figure(figsize=(6,6))
frame = fig.add_subplot(1,1,1)

for folder in os.listdir(ARCiS_path):
    if folder.split("_")[-1] != '5':
        continue
    path = ARCiS_path+folder+"/mixingratios.dat"
    pressure = np.genfromtxt(path,usecols=1,skip_header=1)
    temp = np.genfromtxt(path,usecols=0,skip_header=1)
    frame.plot(temp[::-1],pressure,label=folder)

frame.set_yscale('log')
frame.set_xlabel("Temperature (K)")
frame.set_ylabel("Pressure (bar)")
frame.legend()
show()
```
